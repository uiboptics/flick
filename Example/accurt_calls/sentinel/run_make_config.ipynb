# Generate the Sentinel-3 configuration

This notebook runs `make_config.py` from its own directory and displays the generated `flick_tmp/config` file.

```python
import os
import shutil
import subprocess
from pathlib import Path

project_dir = Path.cwd()
script = project_dir / "make_config.py"
if not script.is_file():
    raise FileNotFoundError(
        f"Could not find {script}. Start Jupyter Lab in the directory containing make_config.py."
    )

# Load Flick from either a source checkout or a binary distribution.
source_root = project_dir.parents[2]
flick_locations = [(source_root / "python_script", source_root)]
flick_locations.extend(
    (parent / "python-package" / "python_script", parent)
    for parent in project_dir.parents
)
try:
    python_module_dir, flick_root = next(
        (module_dir, root)
        for module_dir, root in flick_locations
        if (module_dir / "flick.py").is_file()
    )
except StopIteration:
    searched = "\n  ".join(str(path) for path, _ in flick_locations)
    raise FileNotFoundError(
        f"Could not find the Flick Python module. Searched:\n  {searched}"
    ) from None
os.environ["PYTHONPATH"] = os.pathsep.join(
    [str(python_module_dir), os.environ.get("PYTHONPATH", "")]
)
os.environ["PATH"] = os.pathsep.join(
    [str(flick_root), str(Path.home() / ".local" / "bin"), os.environ.get("PATH", "")]
)
print(f"Working directory: {project_dir}")
print(f"Flick module directory: {python_module_dir}")
print(f"Flick launcher: {shutil.which('flick')}")
print(f"Python executable: {shutil.which('python3')}")
print(f"Running: {script.name}")
```

```python
subprocess.run(["python3", str(script)], cwd=project_dir, check=True)
```

## Generated configuration

```python
config_file = project_dir / "flick_tmp" / "config"
if config_file.is_file():
    print(config_file.read_text())
else:
    print(f"No configuration file found at {config_file}")
```
